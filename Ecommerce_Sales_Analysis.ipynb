# E-commerce Sales Analysis

This notebook analyzes an e-commerce sales dataset using **Python and Pandas**.

### Objectives
- Load and inspect the dataset
- Select and filter data
- Sort orders
- Use `groupby()` and aggregation functions
- Analyze sales by category, city, product, and payment method
- Identify top-performing products/categories and specific orders
- Write meaningful observations


## 1. Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np

# Load the CSV file
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df.head()
```

## 2. Basic Dataset Exploration

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nDescriptive statistics:")
display(df.describe(include="all").T)
```

## 3. Data Selection

```python
# Select a single column
display(df["Product"].head())

# Select multiple columns
display(df[["Order_ID", "City", "Category", "Product", "Quantity", "Total_Sales"]].head())

# Select rows with iloc
display(df.iloc[0:5, 0:6])

# Select rows with loc
display(
    df.loc[df["Total_Sales"] > 10000,
           ["Order_ID", "Product", "Total_Sales"]].head(10)
)
```

## 4. Filtering

```python
# Orders with sales greater than ₹5,000
sales_above_5000 = df[df["Total_Sales"] > 5000]
display(sales_above_5000[["Order_ID", "Product", "Total_Sales"]].head(10))

# Orders where quantity is at least 4
quantity_at_least_4 = df[df["Quantity"] >= 4]
display(quantity_at_least_4[["Order_ID", "Product", "Quantity", "Total_Sales"]].head(10))

# Sports category orders
sports_orders = df[df["Category"] == "Sports"]
display(sports_orders[["Order_ID", "Product", "Quantity", "Total_Sales"]].head(10))

# UPI orders
upi_orders = df[df["Payment_Method"] == "UPI"]
display(upi_orders[["Order_ID", "Product", "Total_Sales", "Payment_Method"]].head(10))

# Combined filter
sports_high_value = df[
    (df["Category"] == "Sports") & (df["Total_Sales"] > 5000)
]
display(sports_high_value[["Order_ID", "Product", "Total_Sales"]])
```

## 5. Sorting

```python
# Highest-value orders
top_orders = df.sort_values("Total_Sales", ascending=False)
display(
    top_orders[["Order_ID", "City", "Category", "Product", "Total_Sales"]].head(10)
)

# Lowest-value orders
lowest_orders = df.sort_values("Total_Sales")
display(
    lowest_orders[["Order_ID", "City", "Category", "Product", "Total_Sales"]].head(10)
)

# Orders with the highest quantity
top_quantity_orders = df.sort_values("Quantity", ascending=False)
display(
    top_quantity_orders[["Order_ID", "Product", "Quantity", "Total_Sales"]].head(10)
)
```

## 6. Overall Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

overall_stats = pd.Series({
    "Total Sales": total_sales,
    "Average Sales per Order": average_sales,
    "Highest Order Sales": highest_sales,
    "Lowest Order Sales": lowest_sales,
    "Total Quantity Sold": total_quantity,
    "Number of Orders": number_of_orders
})

display(overall_stats.to_frame("Value"))
```

## 7. Sales by Category

```python
category_summary = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "count")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(category_summary)
```

## 8. Sales by City

```python
city_summary = (
    df.groupby("City")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "count")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(city_summary)
```

## 9. Product Performance

```python
product_summary = (
    df.groupby("Product")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Number_of_Orders=("Order_ID", "count")
      )
      .sort_values("Total_Sales", ascending=False)
)

print("Top 10 products by total sales:")
display(product_summary.head(10))

print("Top 10 products by quantity sold:")
display(product_summary.sort_values("Total_Quantity", ascending=False).head(10))
```

## 10. Payment Method Analysis

```python
payment_summary = (
    df.groupby("Payment_Method")
      .agg(
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean"),
          Number_of_Orders=("Order_ID", "count")
      )
      .sort_values("Total_Sales", ascending=False)
)

display(payment_summary)
```

## 11. Identifying Specific Orders

```python
# Highest-sales order
highest_order = df.loc[df["Total_Sales"].idxmax()]
print("Highest-sales order:")
display(highest_order.to_frame("Value"))

# Lowest-sales order
lowest_order = df.loc[df["Total_Sales"].idxmin()]
print("Lowest-sales order:")
display(lowest_order.to_frame("Value"))

# Orders with sales between ₹5,000 and ₹10,000
mid_value_orders = df[df["Total_Sales"].between(5000, 10000)]
display(
    mid_value_orders[
        ["Order_ID", "City", "Category", "Product", "Total_Sales"]
    ].sort_values("Total_Sales", ascending=False)
)
```

## 12. Top-Performing Categories and Products

```python
print("Top 3 categories by total sales:")
display(category_summary.head(3))

print("Top 5 products by total sales:")
display(product_summary.head(5))
```

## 13. Meaningful Observations

```python
top_category = category_summary["Total_Sales"].idxmax()
top_city = city_summary["Total_Sales"].idxmax()
top_product = product_summary["Total_Sales"].idxmax()
top_payment = payment_summary["Total_Sales"].idxmax()
top_payment_avg = payment_summary["Average_Sales"].idxmax()

highest_order_id = df.loc[df["Total_Sales"].idxmax(), "Order_ID"]
lowest_order_id = df.loc[df["Total_Sales"].idxmin(), "Order_ID"]

print(f"1. Total sales were ₹{total_sales:,.2f}, with an average order value of ₹{average_sales:,.2f}.")
print(f"2. The highest-value order was {highest_order_id} at ₹{highest_sales:,.2f}.")
print(f"3. The lowest-value order was {lowest_order_id} at ₹{lowest_sales:,.2f}.")
print(f"4. {top_category} was the top category by total sales.")
print(f"5. {top_city} was the top city by total sales.")
print(f"6. {top_product} was the top product by total sales.")
print(f"7. {top_payment} generated the highest total sales among payment methods.")
print(f"8. {top_payment_avg} had the highest average order value.")
print(f"9. A total of {total_quantity} units were sold across {number_of_orders} orders.")
print("10. Sales value, order frequency, and quantity sold should all be considered when evaluating performance.")
```

## 14. Conclusion

The analysis demonstrates how Pandas can be used to transform raw e-commerce transactions into useful business insights. Data selection, filtering, sorting, `groupby()`, and aggregation functions were used to compare performance across categories, cities, products, and payment methods.

These results can help identify strong-performing products and categories, high-value markets, and customer payment preferences.
